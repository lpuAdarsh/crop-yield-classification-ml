```python
# ============================================================
# CROP YIELD PREDICTION USING MACHINE LEARNING
# Complete EDA + Preprocessing + 6 ML Models + Evaluation
# ============================================================


# 1. IMPORT LIBRARIES
# ============================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier, plot_tree
from sklearn.neighbors import KNeighborsRegressor, KNeighborsClassifier
from sklearn.svm import LinearSVC

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    ConfusionMatrixDisplay
)

warnings.filterwarnings('ignore')
sns.set_style("whitegrid")

print("Libraries Imported Successfully!")


# 2. CREATE OUTPUT FOLDERS
# ============================================================

os.makedirs("Crop_Yield_Results", exist_ok=True)
os.makedirs("Crop_Yield_Results/Figures", exist_ok=True)

FIG_DIR = "Crop_Yield_Results/Figures"


def save_plot(filename):
    plt.tight_layout()
    plt.savefig(
        f"{FIG_DIR}/{filename}",
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()


# 3. LOAD DATASET
# ============================================================

df = pd.read_csv("crop_yield.csv")

print("Dataset Loaded Successfully!")

print("Dataset Shape:", df.shape)

display(df.head())


# 4. DATASET UNDERSTANDING
# ============================================================

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nCategorical Summary:")
display(df.describe(include="object"))

print("\nColumn Names:")
print(df.columns.tolist())


# 5. DATA CLEANING
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

df = df.drop_duplicates().copy()

# Standardize categorical values

categorical_cols = [
    'Region',
    'Soil_Type',
    'Crop',
    'Weather_Condition'
]

for col in categorical_cols:
    df[col] = df[col].astype('string').str.strip().str.title()
    df[col] = df[col].replace('', pd.NA)

# Convert numerical columns

numeric_cols = [
    'Rainfall_mm',
    'Temperature_Celsius',
    'Days_to_Harvest',
    'Yield_tons_per_hectare'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Convert Boolean columns

for col in ['Fertilizer_Used', 'Irrigation_Used']:

    if df[col].dtype != bool:

        mapping = {
            'true': True,
            'false': False,
            'yes': True,
            'no': False,
            '1': True,
            '0': False
        }

        df[col] = (
            df[col].astype('string')
            .str.strip()
            .str.lower()
            .map(mapping)
        )

# Negative values are treated as invalid

for col in numeric_cols:
    df.loc[df[col] < 0, col] = np.nan

# Target cannot be missing

df = df.dropna(
    subset=['Yield_tons_per_hectare']
).copy()

print("\nShape After Cleaning:", df.shape)

print("\nRemaining Missing Values:")
print(df.isnull().sum())


# 6. CATEGORICAL ANALYSIS
# ============================================================

print("\nUnique Crops:")
print(df['Crop'].nunique())

print("\nUnique Regions:")
print(df['Region'].nunique())

print("\nUnique Soil Types:")
print(df['Soil_Type'].nunique())

print("\nWeather Conditions:")
print(df['Weather_Condition'].unique())


# 7. EDA - CROP DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x='Crop',
    order=df['Crop'].value_counts().index
)

plt.title("Distribution of Different Crops")
plt.xlabel("Crop")
plt.ylabel("Number of Records")
plt.xticks(rotation=30)

save_plot("Fig1_Crop_Distribution.png")


# 8. REGION DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x='Region',
    order=df['Region'].value_counts().index
)

plt.title("Region-wise Distribution")
plt.xlabel("Region")
plt.ylabel("Count")

save_plot("Region_Distribution.png")


# 9. SOIL TYPE DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x='Soil_Type',
    order=df['Soil_Type'].value_counts().index
)

plt.title("Distribution of Soil Types")
plt.xlabel("Soil Type")
plt.ylabel("Count")
plt.xticks(rotation=25)

save_plot("Soil_Distribution.png")


# 10. YIELD DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    df['Yield_tons_per_hectare'],
    bins=50,
    kde=True
)

plt.title("Distribution of Crop Yield")
plt.xlabel("Yield (Tons per Hectare)")
plt.ylabel("Frequency")

save_plot("Fig2_Yield_Distribution.png")


# 11. RAINFALL DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    df['Rainfall_mm'],
    bins=50,
    kde=True
)

plt.title("Rainfall Distribution")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Frequency")

save_plot("Rainfall_Distribution.png")


# 12. TEMPERATURE DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    df['Temperature_Celsius'],
    bins=40,
    kde=True
)

plt.title("Temperature Distribution")
plt.xlabel("Temperature (Celsius)")

save_plot("Temperature_Distribution.png")


# 13. CROP-WISE YIELD BOXPLOT
# ============================================================

plot_df = df.sample(
    n=min(30000, len(df)),
    random_state=42
)

plt.figure(figsize=(11, 6))

sns.boxplot(
    data=plot_df,
    x='Crop',
    y='Yield_tons_per_hectare'
)

plt.title("Crop-wise Yield Comparison")
plt.xlabel("Crop")
plt.ylabel("Yield (Tons per Hectare)")
plt.xticks(rotation=30)

save_plot("Fig3_Cropwise_Yield_Boxplot.png")


# 14. REGION-WISE YIELD
# ============================================================

plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x='Region',
    y='Yield_tons_per_hectare',
    errorbar=None
)

plt.title("Average Crop Yield by Region")
plt.xlabel("Region")
plt.ylabel("Average Yield")

save_plot("Regionwise_Yield.png")


# 15. RAINFALL VS YIELD
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=plot_df,
    x='Rainfall_mm',
    y='Yield_tons_per_hectare',
    alpha=0.4
)

plt.title("Rainfall vs Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Crop Yield")

save_plot("Fig4_Rainfall_vs_Yield.png")


# 16. TEMPERATURE VS YIELD
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=plot_df,
    x='Temperature_Celsius',
    y='Yield_tons_per_hectare',
    alpha=0.4
)

plt.title("Temperature vs Crop Yield")
plt.xlabel("Temperature")
plt.ylabel("Crop Yield")

save_plot("Temperature_vs_Yield.png")


# 17. WEATHER CONDITION VS YIELD
# ============================================================

plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x='Weather_Condition',
    y='Yield_tons_per_hectare',
    errorbar=None
)

plt.title("Weather Condition vs Crop Yield")
plt.xlabel("Weather Condition")
plt.ylabel("Average Yield")

save_plot("Weather_vs_Yield.png")


# 18. FERTILIZER AND IRRIGATION ANALYSIS
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(11, 5))

sns.barplot(
    data=df,
    x='Fertilizer_Used',
    y='Yield_tons_per_hectare',
    errorbar=None,
    ax=axes[0]
)

axes[0].set_title("Fertilizer Usage vs Yield")

sns.barplot(
    data=df,
    x='Irrigation_Used',
    y='Yield_tons_per_hectare',
    errorbar=None,
    ax=axes[1]
)

axes[1].set_title("Irrigation vs Yield")

save_plot("Fertilizer_Irrigation_Analysis.png")


# 19. CORRELATION HEATMAP
# ============================================================

numeric_df = df.select_dtypes(
    include=np.number
)

plt.figure(figsize=(10, 7))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")

save_plot("Fig5_Correlation_Heatmap.png")


# 20. OUTLIER DETECTION USING IQR
# ============================================================

outlier_columns = [
    'Rainfall_mm',
    'Temperature_Celsius',
    'Days_to_Harvest',
    'Yield_tons_per_hectare'
]

outlier_results = []

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    outlier_results.append({
        'Feature': col,
        'Lower Bound': lower,
        'Upper Bound': upper,
        'Outlier Count': len(outliers)
    })

outlier_df = pd.DataFrame(outlier_results)

display(outlier_df)

outlier_df.to_csv(
    "Crop_Yield_Results/Outlier_Analysis.csv",
    index=False
)


# 21. OUTLIER BOXPLOTS
# ============================================================

for col in outlier_columns:

    plt.figure(figsize=(8, 4))

    sns.boxplot(x=df[col])

    plt.title(f"Outlier Detection - {col}")

    save_plot(f"Boxplot_{col}.png")


# 22. FEATURE AND TARGET SELECTION
# ============================================================

target = 'Yield_tons_per_hectare'

features = [
    'Region',
    'Soil_Type',
    'Crop',
    'Rainfall_mm',
    'Temperature_Celsius',
    'Fertilizer_Used',
    'Irrigation_Used',
    'Weather_Condition',
    'Days_to_Harvest'
]

# Reproducible sample for practical model training

model_df = df.sample(
    n=min(100000, len(df)),
    random_state=42
)

X = model_df[features]
y = model_df[target]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)


# 23. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))


# 24. PREPROCESSING - FIXED
# ============================================================

categorical_features = [
    'Region',
    'Soil_Type',
    'Crop',
    'Weather_Condition'
]

numerical_features = [
    'Rainfall_mm',
    'Temperature_Celsius',
    'Days_to_Harvest'
]

boolean_features = [
    'Fertilizer_Used',
    'Irrigation_Used'
]

# Convert Boolean columns into integer format
for col in boolean_features:
    df[col] = df[col].astype('float64')

# Update X after conversion
X = model_df[features].copy()

for col in boolean_features:
    X[col] = X[col].astype('float64')

# Numerical preprocessing
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Boolean preprocessing
boolean_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent'))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features),
        ('bool', boolean_transformer, boolean_features)
    ],
    sparse_threshold=0
)

# Recreate train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Preprocessing completed successfully!")

# 25. REGRESSION MODELS
# ============================================================

regression_models = {

    'Simple Linear Regression': Pipeline([
        ('preprocessing', preprocessor),
        ('model', LinearRegression())
    ]),

    'Decision Tree Regression': Pipeline([
        ('preprocessing', preprocessor),
        ('model', DecisionTreeRegressor(
            max_depth=15,
            random_state=42
        ))
    ]),

    'KNN Regression': Pipeline([
        ('preprocessing', preprocessor),
        ('model', KNeighborsRegressor(
            n_neighbors=5
        ))
    ])
}


# 26. TRAIN AND EVALUATE REGRESSION
# ============================================================

regression_results = []
fitted_regression = {}

for name, model in regression_models.items():

    print("\nTraining:", name)

    model.fit(X_train, y_train)

    predictions = model.predict(X_test)

    mae = mean_absolute_error(
        y_test,
        predictions
    )

    mse = mean_squared_error(
        y_test,
        predictions
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_test,
        predictions
    )

    regression_results.append({
        'Model': name,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R2 Score': r2
    })

    fitted_regression[name] = model

regression_df = pd.DataFrame(
    regression_results
).sort_values(
    by='R2 Score',
    ascending=False
)

print("\nREGRESSION RESULTS")

display(regression_df)

regression_df.to_csv(
    "Crop_Yield_Results/Regression_Results.csv",
    index=False
)


# 27. REGRESSION MODEL COMPARISON
# ============================================================

plt.figure(figsize=(10, 5))

sns.barplot(
    data=regression_df,
    x='Model',
    y='R2 Score'
)

plt.title("Regression Model Comparison")
plt.xticks(rotation=20)

save_plot("Fig7_Regression_Comparison.png")


# 28. ACTUAL VS PREDICTED
# ============================================================

best_reg_name = regression_df.iloc[0]['Model']

best_reg_model = fitted_regression[best_reg_name]

best_predictions = best_reg_model.predict(X_test)

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    best_predictions,
    alpha=0.4
)

low = min(y_test.min(), best_predictions.min())
high = max(y_test.max(), best_predictions.max())

plt.plot(
    [low, high],
    [low, high],
    linestyle='--'
)

plt.xlabel("Actual Yield")
plt.ylabel("Predicted Yield")

plt.title(f"Actual vs Predicted - {best_reg_name}")

save_plot("Fig6_Actual_vs_Predicted.png")


# 29. CLASSIFICATION TARGET CREATION
# ============================================================

class_df = df.sample(
    n=min(20000, len(df)),
    random_state=42
)

Xc = class_df[features]
yc = class_df[target]

Xc_train, Xc_test, yc_train_raw, yc_test_raw = train_test_split(
    Xc,
    yc,
    test_size=0.20,
    random_state=42
)

# Quantile thresholds calculated from training data only

q1 = yc_train_raw.quantile(1/3)
q2 = yc_train_raw.quantile(2/3)

def classify_yield(values):

    return pd.cut(
        values,
        bins=[-np.inf, q1, q2, np.inf],
        labels=['Low', 'Medium', 'High'],
        include_lowest=True
    ).astype(str)

yc_train = classify_yield(yc_train_raw)
yc_test = classify_yield(yc_test_raw)

print("Training Class Distribution:")
print(yc_train.value_counts())


# 30. CLASSIFICATION MODELS
# ============================================================

classification_models = {

    'Decision Tree': Pipeline([
        ('preprocessing', preprocessor),
        ('model', DecisionTreeClassifier(
            max_depth=15,
            random_state=42
        ))
    ]),

    'KNN': Pipeline([
        ('preprocessing', preprocessor),
        ('model', KNeighborsClassifier(
            n_neighbors=5
        ))
    ]),

    'Linear SVM': Pipeline([
        ('preprocessing', preprocessor),
        ('model', LinearSVC(
            max_iter=5000,
            dual='auto',
            random_state=42
        ))
    ])
}


# 31. TRAIN AND EVALUATE CLASSIFICATION
# ============================================================

classification_results = []
fitted_classification = {}

for name, model in classification_models.items():

    print("\nTraining:", name)

    model.fit(Xc_train, yc_train)

    predictions = model.predict(Xc_test)

    accuracy = accuracy_score(
        yc_test,
        predictions
    )

    precision = precision_score(
        yc_test,
        predictions,
        average='macro',
        zero_division=0
    )

    recall = recall_score(
        yc_test,
        predictions,
        average='macro',
        zero_division=0
    )

    f1 = f1_score(
        yc_test,
        predictions,
        average='macro',
        zero_division=0
    )

    classification_results.append({
        'Model': name,
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1 Score': f1
    })

    fitted_classification[name] = model


classification_df = pd.DataFrame(
    classification_results
).sort_values(
    by='F1 Score',
    ascending=False
)

print("\nCLASSIFICATION RESULTS")

display(classification_df)

classification_df.to_csv(
    "Crop_Yield_Results/Classification_Results.csv",
    index=False
)


# 32. CLASSIFICATION ACCURACY COMPARISON
# ============================================================

plt.figure(figsize=(9, 5))

sns.barplot(
    data=classification_df,
    x='Model',
    y='Accuracy'
)

plt.title("Classification Model Accuracy Comparison")
plt.ylabel("Accuracy")
plt.xticks(rotation=15)

save_plot("Fig8_Classification_Comparison.png")


# 33. CONFUSION MATRIX
# ============================================================

best_clf_name = classification_df.iloc[0]['Model']

best_clf = fitted_classification[best_clf_name]

clf_predictions = best_clf.predict(Xc_test)

print("Best Classifier:", best_clf_name)

print(
    classification_report(
        yc_test,
        clf_predictions,
        zero_division=0
    )
)

fig, ax = plt.subplots(figsize=(7, 6))

ConfusionMatrixDisplay.from_predictions(
    yc_test,
    clf_predictions,
    labels=['Low', 'Medium', 'High'],
    cmap='Blues',
    ax=ax
)

plt.title("Confusion Matrix")

save_plot("Confusion_Matrix.png")


# 34. DECISION TREE VISUALIZATION
# ============================================================

tree_model = fitted_classification[
    'Decision Tree'
].named_steps['model']

feature_names = (
    fitted_classification['Decision Tree']
    .named_steps['preprocessing']
    .get_feature_names_out()
)

plt.figure(figsize=(22, 10))

plot_tree(
    tree_model,
    feature_names=feature_names,
    class_names=list(tree_model.classes_),
    filled=True,
    rounded=True,
    max_depth=3,
    fontsize=7
)

plt.title("Decision Tree Classifier")

save_plot("Decision_Tree.png")


# 35. FEATURE IMPORTANCE
# ============================================================

tree_regressor = fitted_regression[
    'Decision Tree Regression'
].named_steps['model']

feature_names = (
    fitted_regression['Decision Tree Regression']
    .named_steps['preprocessing']
    .get_feature_names_out()
)

importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': tree_regressor.feature_importances_
})

importance_df = importance_df.sort_values(
    by='Importance',
    ascending=False
).head(15)

display(importance_df)

importance_df.to_csv(
    "Crop_Yield_Results/Feature_Importance.csv",
    index=False
)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=importance_df,
    x='Importance',
    y='Feature'
)

plt.title("Top 15 Important Features")

save_plot("Fig9_Feature_Importance.png")


# 36. SAVE CLEANED DATASET
# ============================================================

df.to_csv(
    "Crop_Yield_Results/Cleaned_Crop_Yield.csv",
    index=False
)

print("\nAll Results Saved Successfully!")


# 37. FINAL SUMMARY
# ============================================================

print("\n========== PROJECT SUMMARY ==========")

print("Total Records:", len(df))

print("Total Features:", len(features))

print("\nBest Regression Model:")
print(regression_df.iloc[0])

print("\nBest Classification Model:")
print(classification_df.iloc[0])

print("\nOutput Folder: Crop_Yield_Results")

print("Project Completed Successfully!")
```
